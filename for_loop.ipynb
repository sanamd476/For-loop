```python
# loops has two different types
#1.while
#2.for
#so lets understand for first
#Q1. Print numbers from 1 to 10 using a for loop.
for i in range(1,10+1,1):
  print(i)
```

```python
#2. Range = Range helps to understand start,stop,move like we have to write first four numbers than we write syntax like this:
for i in range(5):
  print(i)
#Q2. Print all odd numbers from 1 to 20 using range().
for i in range(1,20+1,1):
  if i % 2 != 0:
    print(i)
```

```python
#Topic 3: for Loop + if Condition
#Print even numbers
for i in range(1,20+1,1):
  if i % 2 == 0:
    print(i)
```

```python
#Topic 4: for Loop + break
#break = STOP the loop.
#Write a for loop from 1 to 20 and stop the loop when the number reaches 10.
for i in range(1,20+1,1):
  if i == 10:
    break
  print(i)
```

```python
#Topic 5: for Loop + continue
#continue means skip the current iteration and move to the next one.
#Print numbers from 1 to 20, but skip all even numbers.
for i in range(1,20+1,1):
  if i % 2 == 0:
    continue
  print(i)
```

```python
#Topic 6: for Loop + else
#Loop finishes normally → ✅ else runs
#Write a for loop from 1 to 5 that prints each number and then prints:
for i in range(1,5+1,1):
  print(i)
else:
  print("Loop completed")
```

```python
#Topic 7: for Loop with a List
#Print each sale using a for loop.
sales = [100, 250, 400, 150, 500]
for i in range(0,len(sales),1):
  print(sales[i])
```

```python
#Topic 8: for Loop + List + if
sales = [200, 600, 150, 800, 450, 900]
#Print only the sales greater than 500.
for i in range(0,len(sales),1):
  if sales[i]>500:
    print(sales[i])
```

```python
#Topic 9: for Loop + List + Calculations
sales = [100, 200, 300, 400, 500]
#Use a for loop to calculate the total sales.
Total = 0
for i in range(0,len(sales),1):
  Total = Total + sales[i]
print(Total)
```

```python
#Topic 10: for Loop with Strings
name = "Python"
#Use a for loop to print each character separately.
for i in range(0,len(name),1):
  print(name[i])
```

```python
#Topic 11: for Loop with a Tuple
cities = ("Delhi", "Mumbai", "Pune", "Chandigarh")
#Use a for loop to print each city.
for i in range(0,len(cities),1):
  print(cities[i])
```

```python
#Topic 12: for Loop with a Dictionary
employees = {
    "Aman": 45000,
    "Rahul": 65000,
    "Priya": 55000
}

#Write a for loop to print only the employee names.
for i in employees:
  print(i)
```

```python
#Topic 13: Nested for Loops
#A nested loop means a for loop inside another for loop.
#Outer loop → controls the main repetitions
#Inner loop → runs completely for every outer-loop iteration
for i in range(1, 4):
    for j in range(1, 3):
        print("Hello")
```

```python
#Topic 14: enumerate()
#It lets you get index + value together:
cities = ["Delhi", "Mumbai", "Pune"]
for i , city in enumerate(cities):
  print(i,city)
```
